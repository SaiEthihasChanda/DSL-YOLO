```python
# YOLO + GLCM Feature Fusion Pipeline

import os
import cv2
import torch
import numpy as np
import pandas as pd
from skimage.feature import graycomatrix, graycoprops
from ultralytics import YOLO
from tqdm import tqdm

# ----------------------------
# CONFIGURATION
# ----------------------------
TRAIN_IMAGE_DIR = "images/normal/Training_Data"
THERMAL_IMAGE_DIR = "images/Training_Data"
ANNOTATION_DIR = "labels/Training_Data"  # Assuming YOLO format .txt files
SEGMENTED_NORMAL_DIR = "segmented"
SEGMENTED_THERMAL_DIR = "segmented/images/thermal"

os.makedirs(SEGMENTED_NORMAL_DIR, exist_ok=True)
os.makedirs(SEGMENTED_THERMAL_DIR, exist_ok=True)

# ----------------------------
# HELPER FUNCTIONS
# ----------------------------
def read_yolo_annotation(txt_path, img_w, img_h):
    boxes = []
    with open(txt_path, 'r') as f:
        for line in f:
            cls, x_center, y_center, width, height = map(float, line.strip().split())
            x1 = int((x_center - width / 2) * img_w)
            y1 = int((y_center - height / 2) * img_h)
            x2 = int((x_center + width / 2) * img_w)
            y2 = int((y_center + height / 2) * img_h)
            boxes.append((x1, y1, x2, y2))
    return boxes

def crop_and_save(img, boxes, base_filename, save_dir):
    paths = []
    for i, (x1, y1, x2, y2) in enumerate(boxes):
        crop = img[y1:y2, x1:x2]
        seg_path = os.path.join(save_dir, f"{base_filename}_{i}.jpg")
        cv2.imwrite(seg_path, crop)
        paths.append(seg_path)
    return paths

# ----------------------------
# STEP 1 + 2: SEGMENT IMAGES
# ----------------------------
print("Segmenting normal and thermal images...")

segment_map = []
for fname in tqdm(os.listdir(TRAIN_IMAGE_DIR)):
    if not fname.endswith(".jpg"):
        continue
    base_name = os.path.splitext(fname)[0]
    img_path = os.path.join(TRAIN_IMAGE_DIR, fname)
    annotation_path = os.path.join(ANNOTATION_DIR, base_name + ".txt")
    if not os.path.exists(annotation_path):
        continue

    # Read image and get dimensions
    img = cv2.imread(img_path)
    h, w = img.shape[:2]
    boxes = read_yolo_annotation(annotation_path, w, h)

    # Segment normal image
    normal_segments = crop_and_save(img, boxes, base_name, SEGMENTED_NORMAL_DIR)

    # Segment thermal image
    thermal_img_path = os.path.join(THERMAL_IMAGE_DIR, fname)
    thermal_img = cv2.imread(thermal_img_path)
    thermal_segments = crop_and_save(thermal_img, boxes, base_name, SEGMENTED_THERMAL_DIR)

    segment_map.extend(zip(normal_segments, thermal_segments))

# ----------------------------
# STEP 3: GLCM FEATURES
# ----------------------------
def extract_glcm_features(img_path):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    img = cv2.resize(img, (64, 64))  # Resize for consistency
    glcm = graycomatrix(img, distances=[1], angles=[0], levels=256, symmetric=True, normed=True)
    props = ['contrast', 'dissimilarity', 'homogeneity', 'ASM', 'energy', 'correlation']
    feature_vals = [graycoprops(glcm, prop)[0, 0] for prop in props]
    mean = np.mean(img)
    std = np.std(img)
    var = np.var(img)
    min_val = np.min(img)
    max_val = np.max(img)
    entropy = -np.sum(glcm * np.log2(glcm + 1e-10))
    return feature_vals + [mean, std, var, min_val, max_val, entropy]

print("Extracting GLCM features...")
glcm_features = {}
for _, thermal_seg in tqdm(segment_map):
    name = os.path.basename(thermal_seg)
    glcm_features[name] = extract_glcm_features(thermal_seg)

columns_glcm = ['contrast', 'dissimilarity', 'homogeneity', 'ASM', 'energy', 'correlation',
                'mean', 'std', 'var', 'min', 'max', 'entropy']
df_glcm = pd.DataFrame.from_dict(glcm_features, orient='index', columns=columns_glcm)

# ----------------------------
# STEP 4: YOLO FEATURES
# ----------------------------
print("Extracting YOLO features...")
model_yolo = YOLO("runs/detect/train57/weights/best.pt")
layer_to_hook = model_yolo.model.model[-2]  # Use second last layer
features_dict = {}

features = []
def hook(module, input, output):
    features.append(output)

hook_handle = layer_to_hook.register_forward_hook(hook)

for normal_seg, _ in tqdm(segment_map):
    name = os.path.basename(normal_seg)
    img = cv2.imread(normal_seg)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (640, 640))
    img = img.astype(np.float32) / 255.0
    img = torch.from_numpy(img.transpose(2, 0, 1)).unsqueeze(0)
    with torch.no_grad():
        _ = model_yolo.model(img)
    feat = features[0]
    feat = torch.nn.functional.adaptive_avg_pool2d(feat, 1).view(feat.shape[0], -1)
    features_dict[name] = feat.squeeze(0).numpy()
    features.clear()

hook_handle.remove()
df_yolo = pd.DataFrame.from_dict(features_dict, orient='index')

# ----------------------------
# STEP 5: COMBINE
# ----------------------------
print("Combining features...")
df_combined = pd.concat([df_yolo, df_glcm], axis=1)
print(df_combined.shape)

# Save the combined dataframe
# df_combined.to_csv("combined_features.csv")

# Display head
df_combined.head()
```
